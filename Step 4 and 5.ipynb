```python
#step 4 and 5


import numpy as np

n_inc = 1.0
n_sub = 1.5
theta_i_deg = 30
#T_total will be obtained from step 3 here i did example
T_total = np.array([
    [0.98+0.12j, 0,          0.01+0.02j, 0],
    [0,          0.98+0.12j, 0,          0.01+0.02j],
    [0.01+0.02j, 0,          0.98+0.12j, 0],
    [0,          0.01+0.02j, 0,          0.98+0.12j]
], dtype=complex)
theta_i = np.radians(theta_i_deg)

theta_t = np.arcsin(n_inc * np.sin(theta_i) / n_sub)

cos_i = np.cos(theta_i)
cos_t = np.cos(theta_t)


L_inc = np.array([
    [0, cos_i],
    [0, n_inc],
    [1, 0],
    [n_inc * cos_i, 0]
], dtype=complex)


L_ref = np.array([
    [0, -cos_i],
    [0, n_inc],
    [1, 0],
    [-n_inc * cos_i, 0]
], dtype=complex)


L_trans = np.array([
    [0, cos_t],
    [0, n_sub],
    [1, 0],
    [n_sub * cos_t, 0]
], dtype=complex)


T_L = T_total @ L_trans

A = np.hstack((L_ref, -T_L))

incident_s = np.array([[1], [0]])

b_s = -L_inc @ incident_s

x_s = np.linalg.solve(A, b_s)


incident_p = np.array([[0], [1]])

b_p = -L_inc @ incident_p

x_p = np.linalg.solve(A, b_p)


r_ss = x_s[0, 0]
r_ps = x_s[1, 0]

r_sp = x_p[0, 0]
r_pp = x_p[1, 0]


R = np.array([
    [r_ss, r_sp],
    [r_ps, r_pp]
])


print("Matrix A:")
print(A)

print("\nSolution for incident s-polarization:")
print(x_s)

print("\nSolution for incident p-polarization:")
print(x_p)

print("\nReflection Matrix R:")
print(R)
```
